```python
import os
import json
import numpy as np

base_dir = "$PATH/WM_ICV/result_0519_ESNR/resnet"
models = ["t1", "t2", "t1t2", "ratio_origin", "t1t2_ratio_origin"]
models = ["t1", "t2", "t1t2", "ratio_origin"]
num_folds = 5
metric_keys = ["mae"]

print('resnet')
for model in models:
    metrics = {key: [] for key in metric_keys}
    
    for fold in range(num_folds):
        json_path = os.path.join(base_dir, f"{fold}_{model}_15_fsl_None_metrics.json")
        try:
            with open(json_path, "r") as f:
                data = json.load(f)
                for key in metric_keys:
                    metrics[key].append(data[key])
        except FileNotFoundError:
            print(f"file not found: {json_path}")
            continue

    print(f"\n==== Model: {model} ====")
    for key in metric_keys:
        values = np.array(metrics[key])
        if len(values) == 0:
            print(f"{key.upper():<10} No data")
            continue
        mean = np.mean(values)
        std = np.std(values)
        print(f"{key.upper():<10} Mean: {mean:.3f}, Std: {std:.3f}")
```
